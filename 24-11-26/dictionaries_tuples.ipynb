### ex12
נתון ה־Dictionary הבא:

הדפיסו רק תלמידים שקיבלו 80 ומעלה.

ספרו כמה תלמידים נכשלו (מתחת ל־60).

```python
grades = {
    "David": 90,
    "Moshe": 75,
    "Dana": 100,
    "Noa": 55,
    "Avi": 82
}
total = conut_fail = 0
highest = lowest = list(grades.values())[0]
for name, grade in grades.items():
    total += grade
    if grade > highest:
        highest = grade
    if grade < lowest:
        lowest = grade
    if grade < 60:
        conut_fail += 1
    if grade >= 80:
        print(f"Student: {name} | Grade: {grade}")
print("Average:", total / len(grades))
print("Highest:", highest)
print("Lowest:", lowest)
print("Fails:", conut_fail)
```

### ex21
מערכת ניהול אנשי קשר
צרו Dictionary שבו:
Key הוא שם איש הקשר.
Value הוא מספר הטלפון.
הציגו תפריט המאפשר:
הוספת איש קשר.
חיפוש איש קשר.
עדכון מספר טלפון.
מחיקת איש קשר.
הצגת כל אנשי הקשר.
יציאה.
התפריט יפעל באמצעות WHILE.

```python
contacts = {
    "Police": "100",
    "Ambulance": "101",
    "Fire": "102",
    "Electric Corporation": "103",
    "Netivei Israel": "*2120",
    "ZAKA": "1220",
    "United Hatzalah": "1221"
}

is_running = True
while is_running:
    print("""
1. Add Contact
2. Search Contact
3. Update Contact Number
4. Delete Contact
5. Display All Contacts
6. Exit
    """)
    choice = int(input("Choice: "))
    match choice:
        case 1:
            name = input("Add Contact Name: ")
            if name in contacts:
                print(f"{name} Already Exists!")
                continue
            number = input("Add Contact Number: ")
            contacts[name] = number
        case 2:
            name = input("Search Contact By Name: ")
            if name in contacts:
                print(contacts[name])
            else:
                print(f"{name} Not Found")
        case 3:
            name = input("Update Contact By Name: ")
            if name in contacts:
                contacts[name] = input("Enter Updated Number: ")
            else:
                print(f"{name} Not Found")
        case 4:
            name = input("Delete Contact By Name: ")
            if name in contacts:
                contacts.pop(name)
            else:
                print(f"{name} Not Found")
        case 5:
            for name, number in contacts.items():
                print(name, number)
        case 6:
            is_running = False
        case _:
            print("Invalid Choice!")
        
```

# FINAL PROJECT – STORE MANAGEMENT SYSTEM

מטרה: בניית מערכת קטנה לניהול חנות, תוך שימוש ב־Lists וב־Dictionaries.
פירוט הדרישות
Show all products – הצגת כל המוצרים, המחיר והמלאי.
Add product – הוספת מוצר חדש לרשימה.
Search product – חיפוש מוצר לפי שם והצגת פרטיו.
Update product price – שינוי מחיר של מוצר קיים.
Update product stock – שינוי כמות המלאי של מוצר קיים.
Sell product – מכירת מוצר והפחתת הכמות מהמלאי. יש לוודא שיש מספיק מלאי.
Show products by price – הצגת מוצרים שמחירם גבוה ממחיר שהמשתמש יבחר.
Show products with low stock – הצגת מוצרים שכמות המלאי שלהם קטנה מ־10.
Show total inventory value – חישוב שווי המלאי הכולל (מחיר כפול כמות).
Exit – יציאה מהמערכת.
דרישות טכניות
יש להשתמש ב־WHILE עבור התפריט.
יש להשתמש ב־FOR למעבר על רשימת המוצרים.
יש להשתמש ב־DICTIONARY עבור כל מוצר.
יש להשתמש ב־LIST COMPREHENSION לפחות פעם אחת.
אין להשתמש בפונקציות עצמאיות.
אין להשתמש ב־SETS או במבני נתונים נוספים.

```python
products =[
    {
        "name":"Laptop",
        "price":4500,
        "stock":10
    },
    {
        "name":"Mouse",
        "price":80,
        "stock":50
    },
    {
        "name":"Keyboard",
        "price":150,
        "stock":20
    },
    {
        "name":"Screen",
        "price":900,
        "stock":15
    }
]

is_running = True
while is_running:
    print("""
===== Store Management =====

1. Show all products
2. Add product
3. Search product
4. Update product price
5. Update product stock
6. Sell product
7. Show products by price
8. Show products with low stock
9. Show total inventory value
0. Exit

    """)
    choice = int(input("Choice: "))
    match choice:
        case 1:
            for product in products:
                print(f"{product["name"].ljust(20)}Price: {str(product["price"]).ljust(10)}Stock: {product["stock"]}")
        case 2:
            name = input("Enter Product Name: ")
            for product in products:
                if name == product["name"]:
                    print (f"{name} Already Exists!")
                    continue
            price = int(input("Enter Product Price: "))
            quantity = int(input("Enter Product Quantity: "))
            new_product = {
                "name":name,
                "price":price,
                "stock":quantity
            }
            products.append(new_product)
        case 3:
            name = input("Enter Product Name to Search: ")
            for product in products:
                if name == product["name"]:
                    print(f"{product["name"].ljust(20)}Price: {str(product["price"]).ljust(10)}Stock: {product["stock"]}")
                    break
            else:
                print(f"Product {name} Not Found")
        case 4:
            name = input("Enter Product Name to Updsate Its Price: ")
            for product in products:
                if name == product["name"]:
                    price = int(input("Enter New Product Price: "))
                    product["price"] = price
                    break
            else:
                print(f"Product {name} Not Found")
        case 5:
            name = input("Enter Product Name to Updsate Its Stock: ")
            for product in products:
                if name == product["name"]:
                    stock = int(input("Enter New Product Stock: "))
                    product["stock"] = stock
                    break
            else:
                print(f"Product {name} Not Found")
        case 6:
            name = input("Enter Product Name to Buy: ")
            for product in products:
                if name == product["name"]:
                    quantity = int(input("Enter Quantity: "))
                    if quantity > product["stock"]:
                        print("Not Enough In Stock")
                        break
                    else:
                        product["stock"] -= quantity
            else:
                print(f"Product {name} Not Found")
        case 7:
            price_threshold = int(input("Enter Price threshold: "))
            for product in products:
                if product["price"] >= price_threshold:
                    print(f"{product["name"].ljust(20)}Price: {str(product["price"]).ljust(10)}Stock: {product["stock"]}")
        case 8:
            for product in products:
                if product["stock"] < 10:
                    print(f"{product["name"].ljust(20)}Price: {str(product["price"]).ljust(10)}Stock: {product["stock"]}")
        case 9:
            total = sum(product["price"] for product in products)
            print("Total:", total)
        case 0:
            is_running = False
        case _:
            print("Invalid Choice!")
```
